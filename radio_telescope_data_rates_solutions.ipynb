## Exploring Radio Telescope Data Rates

This notebook lets you explore how data volume scales with different telescope parameters. We refer to our data products as visibilities. These are correlations between the electric fields measured at different antennas.

The data rate depends on the following parameters:
- $N_{annenta}$: the number of antennae
- $N_{freq}$: the number of frequency channels
- dump_time: how often do you want to save a correlation
- $N_{corr}$: the number of correlations
- bytes_per_correlation: how many bytes does it take to store one correlation?

### Number of antennas

Including more and more antennas in your interferometers helps:
- increase the sensitivity of your telescope
- increase the resolution of your observations
- increase the telescope's ability to measure extended structures

It is not directly the number of antennas that impacts the data rate, but the number of correlations we can obtain from those antennas. These correlations are formed by correlating singals of antenna pairs. We often refer to antenna pairs as **baselines**.

```python
def number_of_baselines(N_antenna):
    """Convert the number of antennas in your array to a number of unique baselines."""
    return N_antenna*(N_antenna-1)/2
```

### Number of frequency channels

Extremely high spectral resolution is required to measure the spectral line width of different (molecular) absorption/emission lines.


### Dump time

How often do you want to save a visibility? For transient science you might want go to 1 second or even lower. For other science cases this resolution might not be necessary, but we often still store visibilities every 10 seconds, and average afterwards. 

### Number of correlations

The number of correlations you store impact the polarisation capabilities of your observation. Most modern interferometers store all four correlations.
In case you are not interested in the polaristaion of your source, or you expect there to be no polarised signal, you might opt to only store two correlations.
- 2 correlations: “How bright is the signal in two directions?”
- 4 correlations: “How is the electric field oriented and evolving?”

### Bytes per correlation


A single correlation consists of:
- the complex visibility with a real and imaginary part, 2 x float32 = 2x4=8 bytes
- a weight: float32 = 4 bytes
- a flag: true/false = 1-4 bytes (depending on telescope)

```python
def data_rate(N_antenna=100, N_freq=1000,  dump_time=1.0, N_corr=4, bytes_per_visibility=16):
    """Function that calulates the data rate of an interferometer in TeraByte per day."""
    
    # Number of baselines
    N_baselines = number_of_baselines(N_antenna) # complete the number_of_baselines function first

    # Visibilities per second
    N_time = 1 / dump_time

    total_vis_per_sec = N_time * N_baselines * N_freq * N_corr
    data_rate_bytes = total_vis_per_sec * bytes_per_visibility

    # conversions
    MBps = data_rate_bytes / 1e6
    GBps = data_rate_bytes / 1e9
    TB_day = data_rate_bytes * 86400 / 1e12
    
    print(f'Data rate: {MBps:.2f} MB/s ({GBps:.3f} GB/s)')
    print(f'Data per day: {TB_day:.2f} TB/day')
    return TB_day
```
